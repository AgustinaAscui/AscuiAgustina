```python
#PROBLEMA NUMERO 1

print(list(range(-30,51)))
# El primer print es para imprimir el rago de números que van del -30 al 50

for numb in range(-30, 51): #Esta linea indica un ciclo for en donde la variable numb tomara los valores del rango uno por uno

    if numb < 0: #Esta linea inicia un condicional if en el que numb tiene que ser menor que 0
        perception = "Muy frío" #Esta linea declara que perception toma el valor del texto "Muy frío"
        print("La temperatura es de:",numb,"grados por tanto está",perception) #Esta linea imprime en consola el valor de numb y de percepcion junto a un texto coherente a los resultados de la condición

    elif numb >= 0 and numb <= 24:#Esta linea inicia un condicional elif en el que numb tine que tomar el valor entre 0 al 24
        perception = "Templado"
        print("La temperatura es de:",numb,"grados por tanto está",perception)

    else:#Esta linea inicia una condicion mientras no se cumplan las otros condicionales if, es decir el rango que queda del 25 en adelante
        perception = "Caluroso"
        print("La temperatura es de:",numb,"grados por tanto está",perception)
```

```python
#PROBLEMA NUMERO 2

for numb in range(1, 101):
    # Esta línea crea un ciclo for donde numb recorrerá los números del 1 al 100

    if numb % 10 == 0:
        # Esta condición revisa si el número es divisible por 10

        sonido = "Bip Bip"
        # La variable sonido toma el valor "Bip Bip"

        print("El número", numb, "emite el sonido", sonido)
        # Esta línea imprime el número junto a su sonido correspondiente

    elif numb % 5 == 0:
        # Esta condición revisa si el número es divisible por 5

        sonido = "Bip"
        # La variable sonido toma el valor "Bip"

        print("El número", numb, "emite el sonido", sonido)
        # Esta línea imprime el resultado en consola
```

```python
#PROBLEMA NUMERO 3

puntaje1 = int(input("Ingresa el primer puntaje: "))
# Esta línea solicita al usuario ingresar el primer puntaje y lo convierte en número entero

puntaje2 = int(input("Ingresa el segundo puntaje: "))
# Esta línea solicita el segundo puntaje y también lo convierte en entero

suma = puntaje1 + puntaje2
# Esta línea suma ambos puntajes y guarda el resultado en la variable suma

print("La suma total de los puntajes es:", suma)
# Esta línea imprime el resultado total de la suma

if suma < 50:
    # Esta condición revisa si la suma es menor que 50

    nivel = "Nivel principiante"
    # La variable nivel toma el valor "Nivel principiante"

    print("El jugador obtuvo:", nivel)
    # Esta línea imprime el nivel correspondiente

elif suma >= 50 and suma < 80:
    # Esta condición revisa si la suma está entre 50 y 79

    nivel = "Nivel intermedio"
    # La variable nivel toma el valor "Nivel intermedio"

    print("El jugador obtuvo:", nivel)
    # Esta línea imprime el nivel correspondiente

else:
    # Esta condición se ejecuta cuando la suma es 80 o mayor

    nivel = "Nivel pro"
    # La variable nivel toma el valor "Nivel pro"

    print("El jugador obtuvo:", nivel)
    # Esta línea imprime el resultado final
```
